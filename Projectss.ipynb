##  Identify the significant variables by performing statistical analysis using p-values

```python
from scipy.stats import f_oneway

groups = [group['Time_in_seconds'].dropna().values 
          for name, group in data.groupby('Complaint Type')]

f_stat, p_value = f_oneway(*groups)

print("P-value:", p_value)
```

## Perform a Kruskal-WallisH test

```python
from scipy.stats import kruskal

# Create groups based on complaint type
groups = [
    group['Time_in_seconds'].dropna().values
    for name, group in data.groupby('Complaint Type')
]

# Apply test
h_stat, p_value = kruskal(*groups)

print("H-statistic:", h_stat)
print("P-value:", p_value)
```

## Fail to reject H0: All sample distributions are equal
## Reject H0: One or more sample distributions are not equal

```python
if p_value > 0.05:
    print("Fail to reject H0")
elif p_value < 0.05:
    print("Reject H0")
```
